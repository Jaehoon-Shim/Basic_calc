```python
import torch
import torch.nn as nn

# 데이터 2개, 출력 2개: 크기 [N, n_out] = [2, 2]
pred = torch.tensor([
    [1.0, 3.0],  # 데이터 A의 예측값
    [2.0, 2.0],  # 데이터 B의 예측값
])

target = torch.tensor([
    [0.0, 0.0],  # 데이터 A의 정답
    [0.0, 0.0],  # 데이터 B의 정답
])

# 원소별 제곱오차는 다음과 같음
# [[1, 9],
#  [4, 4]]
```

```python
# ① 데이터별 출력 평균 → 크기 [N]
# dim=1: 각 행에서 출력들을 평균
# [(1 + 9) / 2, (4 + 4) / 2] = [5, 4]
loss1 = nn.MSELoss(reduction="none")(pred, target).mean(dim=1)
print("① 데이터별 출력 평균:", loss1.tolist())  # [5.0, 4.0]
```

```python

# ② 출력은 합산하고, 데이터를 평균 → 스칼라
# sum(dim=1): 각 행의 합 [10, 8]
# mean(): 두 데이터의 평균 (10 + 8) / 2 = 9
loss2 = nn.MSELoss(reduction="none")(pred, target).sum(dim=1).mean()
print("② 출력 합산 후 데이터 평균:", loss2.item())  # 9.0
```

```python

# ③ 모든 원소의 평균: PyTorch 기본 MSE → 스칼라
# (1 + 9 + 4 + 4) / 4 = 4.5
loss3 = nn.MSELoss()(pred, target)
print("③ PyTorch 기본 MSE:", loss3.item())  # 4.5
```
